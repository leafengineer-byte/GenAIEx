# Experiment 8: Text-to-Music Generation using pretrained MusicGen
**Aim:** Generate music from text prompts using a pretrained foundation model, play it, and save the audio. No dataset collection or model training is needed.

> GPU is recommended. The first run downloads pretrained weights.

```
import torch
print('CUDA available:', torch.cuda.is_available())
if torch.cuda.is_available(): print(torch.cuda.get_device_name(0))
```

## 1. Load pretrained MusicGen model

```
from audiocraft.models import MusicGen
from audiocraft.data.audio import audio_write
from IPython.display import Audio, display
from pathlib import Path
model_name = 'facebook/musicgen-small'
device = 'cuda' if torch.cuda.is_available() else 'cpu'
model = MusicGen.get_pretrained(model_name, device=device)
model.set_generation_params(duration=8)
print('Model loaded on', device)
```

## 2. Enter a prompt and generate music

```
prompt = 'Peaceful instrumental piano with soft flute, warm ambient textures, slow tempo'
model.set_generation_params(duration=8)
with torch.no_grad(): wav = model.generate([prompt], progress=True)
print('Prompt:', prompt, '| Audio tensor shape:', tuple(wav.shape))
```

## 3. Play generated audio

```
audio = wav[0].detach().cpu()
display(Audio(audio.numpy(), rate=model.sample_rate))
```

## 4. Save audio as WAV and download it

```
out_dir = Path('outputs'); out_dir.mkdir(exist_ok=True)
out_path = out_dir / 'musicgen_sample'
audio_write(str(out_path), audio, model.sample_rate, strategy='loudness')
print('Saved:', str(out_path) + '.wav')
from google.colab import files as colab_files
colab_files.download(str(out_path) + '.wav')
```

## 5. Try another prompt (optional)

```
prompt2 = 'Upbeat electronic instrumental music with bright synthesizers, energetic rhythm, cinematic build-up'
model.set_generation_params(duration=8)
with torch.no_grad(): wav2 = model.generate([prompt2], progress=True)
display(Audio(wav2[0].detach().cpu().numpy(), rate=model.sample_rate))
audio_write('outputs/musicgen_sample_2', wav2[0].detach().cpu(), model.sample_rate, strategy='loudness')
print('Saved outputs/musicgen_sample_2.wav')
```
